```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df=pd.read_csv('hd.csv')
```

```python
df.head()
```

```python
X=df['Head Size(cm^3)'].values
Y=df['Brain Weight(grams)'].values
```

```python
np.corrcoef(X,Y)
#array([[1.    ,0.79956971],[0.79956971,1.    ]])
```

plt.scatter(X,Y,c='green',label='Data points')
plt.xlabel('head size in grams')
plt.ylabel('Brain Weight in grams')
plt.legend()
plt.show()

```python
mean_x=np.mean(X)
mean_y=np.mean(Y)
```

```python
n = len(X)

numer=0
denom=0
for i in range(n):
    numer += (X[i] - mean_x) * (Y[i] - mean_y)
    denom += (X[i] - mean_x) ** 2
b1 = numer/denom
b0 = mean_y - (b1 * mean_x)

print("coefficients for regression",b1,b0)
```

```python
%matplotlib inline

plt.rcParams['figure.figsize'] = (10.0,5.0)

y = b0 + b1 * X

plt.plot(X,y,color='blue',label='Regression Line')
plt.scatter(X,Y,c='green',label='Scatter data')

plt.xlabel('Head Size in cm3')
plt.ylabel('Brain Weight in grams')
plt.legend()
plt.show()
```
