```python
import pandas as pd
import numpy as np
df = pd.read_csv('data.csv')
df.head(9)
```

```python
df.tail()
```

```python
print("missing values : ")
print(df.isnull().sum(),"\n")
```

```python
df_dropped_cols = df.dropna(axis=1)
print("dropped cols \n",df_dropped_cols)
```

```python
df_pandas_filled = df.copy()
df_pandas_filled["Age"] = df_pandas_filled["Age"].fillna(df_pandas_filled["Age"].mean())
df_pandas_filled.head()
```

```python
df_pandas_filled1 = df.copy()
df_pandas_filled1["Salary"] = df_pandas_filled1["Salary"].fillna(df_pandas_filled1["Salary"].median())
df_pandas_filled1.head(10)
```

```python
df_pandas_filled1["Age"] = df_pandas_filled1["Age"].fillna(df_pandas_filled1["Age"].mean())
df_pandas_filled1.head(10)
```

```python
df_pandas_filled1["Department"] = df_pandas_filled1["Department"].fillna(df_pandas_filled1["Department"].mode()[0])
df_pandas_filled1.head(10)
```
