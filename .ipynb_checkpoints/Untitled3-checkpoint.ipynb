```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
dataset=pd.read_csv("Melbourne_housing_FULL.csv")
```

```python
dataset.shape
```

```python
dataset.head()
```

```python
dataset.nunique()
```

```python
cols_use = [ 'Suburb', 'Rooms', 'Type', 'Method', 'SellerG', 'Regionname', 'Propertycount', 'Distance', 'CouncilArea', 'Bedroom2', 'Bathroom', 'Car', 'Landsize', 'BuildingArea', 'Price' ]
```

```python
dataset.shape
```

```python
dataset.isnull().sum()
```

```python

cols_zero = [ 'Propertycount', 'Distance', 'Bedroom2', 'Bathroom', 'Car' ]
dataset[cols_zero] = dataset[cols_zero].fillna(0)
dataset['Landsize'] = dataset['Landsize'].fillna(dataset['Landsize'].mean())
dataset['BuildingArea'] = dataset['BuildingArea'].fillna(dataset['BuildingArea'].mean())
```

```python
dataset.dropna(inplace=True)
```

```python
dataset.shape
```

```python
dataset = pd.get_dummies(dataset, drop_first=True)
```

```python
dataset.head()
```

```python
X = dataset.drop('Price', axis=1)
y = dataset['Price']
```

```python
from sklearn.model_selection import train_test_split
train_X, test_X, train_y, test_y = train_test_split( X, y, test_size=0.3, random_state=2 )
```

```python
from sklearn.linear_model import LinearRegression
reg = LinearRegression()
reg.fit(train_X, train_y)
```

```python
reg.score(train_X, train_y)
```

```python
reg.score(test_X, test_y)
```
