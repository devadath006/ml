```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn import linear_model
```

```python
df=pd.read_csv('house_price.csv')
df.head()
```

```python
X=df['Area'].values
Y=df['Price'].values
np.corrcoef(X,Y)
```

```python
mean_x=np.mean(X)
mean_y=np.mean(Y)
n = len(X)

numer=0
denom=0
for i in range(n):
    numer += (X[i] - mean_x) * (Y[i] - mean_y)
    denom += (X[i] - mean_x) ** 2
b1 = numer/denom
b0 = mean_y - (b1 * mean_x)

print("coefficients for regression",b1,b0)
```

```python
%matplotlib inline

plt.rcParams['figure.figsize'] = (10.0,5.0)

y = b0 + b1 * X

plt.plot(X,y,color='blue',label='Regression Line')
plt.scatter(X,Y,c='green',label='Scatter data')

plt.xlabel('area')
plt.ylabel('price')
plt.legend()
plt.show()
```

```python
new_df = df.drop('Price', axis='columns')
new_df.head()
```

```python
reg = linear_model.LinearRegression()
reg.fit(new_df,Y)
```

```python

reg.predict([[3300]])
```
